
- Asks user what currency they would like to convert into PLN(e.g. CHF).
- Asks user about amount they would like to convert.
- Checks if amount is a number
- Give your APIKEY
 

````python
currency = input("Into which currency would you like to convert PLN(e.g. CHF): ")

try:
    amount = float(input("How much PLN would you like to convert: "))
except Exception:
    print("Amount has to by a number")

APIKEY = # APIKEY from https://exchangeratesapi.io/
````

- Using `requests` library asks about  current exchange rate of the given currency to PLN.
- checks if currency is in data

````python
import requests

response = requests.get(f"http://api.exchangeratesapi.io/latest?access_key={APIKEY}")

if response.status_code == 200:
    data = response.json()
else:
    print("There's no such currency")
````

Conversion to PLN and provides the results in form:

```
<GIVEN AMOUNT> <GIVEN CURRENCY> = <CALCULATED AMOUN> <PLN>

````python
EURO_to_currency = data['rates'][currency] # euro to given currency exchange rate
EURO_to_PLN = data['rates']['PLN'] # euro to PLN exchange rate
currency_to_PLN = round(((EURO_to_PLN / EURO_to_currency) * amount), 2)    


print(f"{amount} {currency} = {currency_to_PLN} PLN")
````
