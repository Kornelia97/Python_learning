Suggests a random activity you can do to kill your boredom . Based on the API from [https://bored-api.appbrewery.com/]

```python
import requests
import random

friends = int(input("How many friends do you want to meet: "))

participants = friends + 1  # Including yourself

response = requests.get(f"https://bored-api.appbrewery.com/filter?participants={participants}")

if response.status_code == 200:
    data = response.json()
    random_activity = random.choice(data)
    print(f"Suggested activity for {participants} people: {random_activity['activity']}")
else:
    print(f"No activities found for {participants} participants.")

```
