The `get_weather` function, which takes three arguments:
- `city` - the name of the city, **required**,
- `state` - the name of the state, optional(works only for USA),
- `country` - the name of the country, optional.

The function returns a short name for the weather.

 Remember to provide `appid` from [openweathermap.org](openweathermap.org). 


```python
import requests

def get_weather(city, state = None, country = None):
   API_key = # ### Provide here API key from OpenWeatherMap
   response = requests.get(f"https://api.openweathermap.org/data/2.5/weather?q={city},{country}&appid={API_key}")
   data = response.json()
   return data['weather'][0]['description']

```

```python
print('Pogoda dla Warszawy:', get_weather('Warszawa'))
print('Warszawa jeszcze raz:', get_weather('Warszawa', 'Polska'))
print('Pogoda dla Nowego Jorku:', get_weather('New York', 'new York', 'USA'))
```
