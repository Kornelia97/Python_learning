Downloads from [https://exchangeratesapi.io/](https://exchangeratesapi.io/) current exchange rates, and displays them as a list:

```
- <Exchange rate 1> EUR/<currency 1>
- <Exchange rate 2> EUR/<currency 2>
...
```

Endpoint /latest - Fetch the most recent exchange rates (timestamped), default base is EUR unless overridden

````python
import requests

APIKEY = # "YOUR_API_KEY" from https://exchangeratesapi.io/
response = requests.get(f"http://api.exchangeratesapi.io/latest?access_key={APIKEY}")
data = response.json()
````

````python
print(data['rates'])
````

````python
for i in data['rates']:

    print(f"- {data['rates'][i]} EUR/{i}")
````
