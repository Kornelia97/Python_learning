Download current exchange rates returned by the API at the address [http://api.nbp.pl/api/exchangerates/tables/a/](http://api.nbp.pl/api/exchangerates/tables/a/), and  display them one below the other, in the following format:

```
AUD (dolar australijski): 2.8571 PLN
USD (dolar amerykański): 3.8849 PLN
```

````python
import requests
import json

response = requests.get("http://api.nbp.pl/api/exchangerates/tables/a/")

data = response.json()
````

````python

for i in data[0]['rates']:
    print(f"{i['code']} ({i['currency']}): {i['mid']}PLN")
````
