Checking password

```python
password = input("Enter your password: ")

if len(password) < 5:
    print("Password is too short. It must be at least 5 characters long.")
elif len(password) > 10:
    print("Password is too long. It must be no more than 10 characters long.")
else:
    pass
```

for loop

```python
for i in range(0, 10, 2):
    print(i)
```

```python
for i in range(10, 0, -1):
    print(i)
```

```python
for i in range(11):
    print(f"2^{i} = {2**i}")
```

```python
numbers = [1, 3, 5]

for i in numbers:
    if i % 2 == 0:
        print("Even number found:", i)
        break
else:
        print("There are no even numbers in the list.")
```

```python
sum_even = 0
for i in range (10):
    if i % 2 :
        print(f"i = {i} Skip odd numbers")
        continue

    print(f"i = {i}")
    sum_even += i

print(sum_even)
```

While loop

```python
import random 

i = 0
while i < 3:
    print(i)
    if random.randint(1,10) == 5:
        print("Break")
        break
    i += 1
```
