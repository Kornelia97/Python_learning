Function `find_first_duplicate(text)`returns first letter that appears more than once from its position to the end of the string. If no letters repeat, the function should return `None`.


```python
def count_letters(text):
    dict = {}
    for letter in text:
        if letter not in dict:
            dict[letter] = 1
        else:
            dict[letter] += 1
    return dict 

count_letters("Hello World")
```

```python
# 1. solution

def find_first_duplicate(text):
    letter_counts = count_letters(text)
    for i in letter_counts:
        if letter_counts[i] > 1:
            return i
    
```

```python
# 2. solution

def find_first_duplicate(text):
    for letter in text:
        if text.count(letter) > 1:  
            return letter
```

```python
if find_first_duplicate('procesor') == 'r':
    print('Pass!')
else:
    print("Function should return 'r' for 'procesor'")
    print('Your function returned:', find_first_duplicate('procesor'))

if find_first_duplicate('klawiatura') == 'a':
    print('Pass!')
else:
    print("Function should return 'a' for 'klawiatura'")
    print('Your function returned:', find_first_duplicate('klawiatura'))

if find_first_duplicate('mysz') == None:
    print('Pass!')
else:
    print("Function should return None for 'mysz'")
    print('Your function returned:', find_first_duplicate('mysz'))
```
