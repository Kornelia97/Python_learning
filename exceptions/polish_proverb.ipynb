Funkcja `proverb()` miała znać przysłowie na każdą sekundę.
Niestety, zna ich tylko siedem, więc przez większość czasu rzuca wyjątek `KeyError`.

Funkcję `proverb_safe()` **użyje** funkcji `proverb()` do pobrania przysłowia na daną sekundę; gdy się nie uda - powinna **złapać wyjątek** i zamiast niego **zwrócić** string `"Brak przysłowia"`.


```python
import datetime

def proverb():
    phrases = {
        0: 'Apetyt rośnie w miarę jedzenia',
        4: 'Co dwie głowy, to nie jedna',
        12: 'Kto pyta, nie błądzi',
        13: 'Raz na wozie raz pod wozem',
        35: 'Gdyby kózka nie skakała, toby nóżki nie złamała',
        44: 'Wybierać się jak sójka za morze',
        59: 'Jedna jaskółka wiosny nie czyni'
    }
    return phrases[datetime.datetime.now().second]
```

```python
def proverb_safe():
    try:
         return proverb()
    except Exception:
         return("Brak przysłowia")
         
```

```python
from unittest.mock import patch

with patch('datetime.datetime') as dt:
    dt.now().second = 40
    if proverb_safe() == 'Brak przysłowia':
        print('Zaliczone!')
    else:
        print('Oczekiwano, że dla 40. sekundy, funkcja proverb_safe zwróci: "Brak przysłowia"')
        print('Funkcja zwróciła:', proverb_safe())

with patch('datetime.datetime') as dt:
    dt.now().second = 44
    if proverb_safe() == 'Wybierać się jak sójka za morze':
        print('Zaliczone!')
    else:
        print('Oczekiwano, że dla 44. sekundy, funkcja proverb_safe zwróci: "Wybierać się jak sójka za morze"')
        print('Funkcja zwróciła:', proverb_safe())

with patch('datetime.datetime') as dt:
    dt.now().second = 50
    if proverb_safe() == 'Brak przysłowia':
        print('Zaliczone!')
    else:
        print('Oczekiwano, że dla 50. sekundy, funkcja proverb_safe zwróci: "Brak przysłowia"')
        print('Funkcja zwróciła:', proverb_safe())
```
