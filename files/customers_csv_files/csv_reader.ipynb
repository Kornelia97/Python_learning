W folderze `archive/` znajdziesz foldery `2018`, `2019` oraz `2020`,a w każdym z nich foldery nazwane trzyliterowymi skrótami angielskich miesięcy (`jan`, `feb`, etc.).

W każdym z tych folderów znajdziesz pliki `.csv`, które zawierają informacje o tym, ilu nowych klientów, stałych klientów oraz klientów VIP odwiedziło tego dnia sklep.

Napisz skrypt, który podsumuje, ilu klientów nowych, stałych oraz VIP odwiedziło sklep w roku 2018, 2019 oraz 2020.


```python
import glob
import csv


#recursive=True: Tells the function to search all nested subdirectories. 
# (By default, recursive is False, and ** only acts like a single *)
def files_in_years(year):
    file = glob.glob(f"archive/{year}/**/*.csv", recursive = True)
    return file
```

```python
for i in range (2018, 2021):
    file_names = files_in_years(i)

    new = 0
    returning = 0
    vip = 0

    for file_name in file_names:
        with open(file_name, newline='') as my_file:
            reader = csv.reader(my_file)
            data = list(reader)
            new += int(data[0][1])
            returning += int(data[1][1])
            vip += int(data[2][1])
    print(f"W roku {i} sklep odwiedziło: ")
    print(f"- {new} nowych klientów")
    print(f"- {returning} powracających klientów")
    print(f"- {vip} klientów VIP.")
```
