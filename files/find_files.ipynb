Znajdź wszystkie pliki z rozszerzeniem **.txt** w folderze `data/`, w pętli otwórz każdy z nich, i wyświetl jego treść na ekranie.

```python
import glob

files = glob.glob(r'data/*txt')

for i in files:
    with open(i) as data:
        for i in data:
            print(i)
```
