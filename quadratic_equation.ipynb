a * x ** 2 + b * x + c == 0

```python
a = float(input("a: "))
b = float(input("b: "))
c = float(input("c: "))
```

```python
discriminant = b ** 2 - 4 * a * c

print(discriminant)
```

Roots

```python
if discriminant > 0:
    x_1 = (-b + discriminant ** 0.5) / (2 * a)
    x_2 = (-b - discriminant ** 0.5) / (2 * a)

    print(f"x_1 = {x_1}")
    print(f"x_12 = {x_2}")
elif discriminant == 0:
    x = - b / ((2 * a))

    print(f"x_1 = x_2 = {x}")
else:
    print("There are no real roots to this equation.")
```
