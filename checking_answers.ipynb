Problem

Ostatnio zwrócił się do nas z prośbą o pomoc nasz przyjaciel — nauczyciel angielskiego w szkole. Spędza on bardzo dużo czasu na ocenie testów, które robi w swoich klasach. Wpadł na pomysł, że część testów, w których są pytania wielokrotnego wyboru, można spróbować oceniać automatycznie. Poprosił nas więc o pomoc w realizacji tego pomysłu.

Do naszego programu będziemy otrzymywać następujące dane:

    Listę z poprawnymi odpowiedziami,
    Listę z odpowiedziami jednego z uczniów.

Nasz przyjaciel spodziewa się, że program wyświetli mu informacje o tym, ile odpowiedzi poprawnie oddał uczeń.

```python
key_answers = ['a', 'b', 'a', 'c', 'd', 'a', 'c', 'a', 'b']
student_answers = ['b', 'b', 'a', 'a', 'd', 'a', 'd', 'a', 'b']

good_answers = 0;
answers_length = len(key_answers)

for i in range(answers_length):
    if(key_answers[i] == student_answers[i]):
        good_answers = good_answers + 1

print(f"Checked {i+1} question")

print(f"Student got {good_answers} answers right out of {answers_length} questions")
```
