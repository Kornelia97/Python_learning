Write a program that:

- Stores a set of numbers in a list (in a variable named `numbers`)
- Calculates their sum and average
- Displays these numbers on the screen
- Compares them against a specific structure, and provides a summary of the comparison results.


```python
numbers = []
j = input("Press 1 if you know how many numbers you want to enter, else press any other key.")
if j == '1':
    hm_numbers = int(input("How many numbers do you want?"))
    for i in range(hm_numbers):
        number = input("Enter number: ")
        numbers.append(number)
else:
    i = 'y'
    while i == 'y':
        number = input("Enter number: ")
        numbers.append(number)
        i = input("if you want more numbers enter 'y'")
print(f"Entered numbers: {', '.join(numbers)}")

sum_num = 0
for num in numbers:
    sum_num += float(num)
    avg_num = sum_num / len(numbers)
print(f"sum: {sum_num} \naverage= {avg_num}")

if sum_num > avg_num:
    print("Sum is bigger")
elif sum_num < avg_num:
    print("Average is bigger")
else:
    print("Average and sum are equal")
```
