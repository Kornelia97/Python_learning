```python
numbers = [1, 2, 3]
print(numbers)

numbers.append(4)
print(numbers)

numbers.extend([5, 6])  # equivalent to `numbers + [5, 6]`
print(numbers)

numbers.insert(2, 42)  # insert number 42 on index 2; rest of the numbers moves one right
print(numbers)
```
