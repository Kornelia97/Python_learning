Function **count_letters** returns a dictionary that counts the number of occurrences of each letter.


```python
def count_letters(text):
    dict = {}
    for letter in text:
        if letter not in dict:
            dict[letter] = 1
        else:
            dict[letter] += 1
    return dict 

count_letters("Hello World")
```

```python
if count_letters('Anna') == {'A': 1, 'n': 2, 'a': 1}:
    print('Pass!')
else:
    print("For the argument 'Anna' function should returned {'A': 1, 'n': 2, 'a': 1} ")
    print('Your function returned: ', count_letters('Anna'))

if count_letters('Kajak') == {'K': 1, 'a': 2, 'j': 1, 'k': 1}:
    print('Pass!')
else:
    print("For the argument 'Kajak' function should returned {'K': 1, 'a': 2, 'j': 1, 'k': 1} ")
    print('Your function returned: ', count_letters('Kajak'))
```
