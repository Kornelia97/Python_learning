Count every natural number from 0 to n.

```python
# First solution

n = int(input("n: "))
sum_n = 0

for i in range(n+1):
    sum_n += i
    print(sum_n)
```

```python
# Second solution

n = int(input("n: "))
sum_n = 0
i = 0

while i <= n:
    sum_n += i
    i += 1
    print(sum_n)
```

```python
# Third solution

n = int(input("n: "))

sum_n = sum(range(0, n+1))

print(sum_n)
```
