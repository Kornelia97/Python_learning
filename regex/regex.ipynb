Regex finds phone numbers

```python
import re
 
example_text = """Julianna Czerkies +48 889156078
Mieszko Łaciak +48 604947843
Adrian Piotrkowicz +48 791640996
Paweł Szałas +48 516111667
Dorota Woda +48 735606830"""
 
regex = r'\+\d{1,3} \d{9}'
 
found = re.findall(regex, example_text)
print('Znaleziono', len(found), 'numerów telefonów:')
print(found)
```

```python
regex = r"(?:[a-z0-9!#$%&'*+\/=?^_`{|}~-]+(?:\.[a-z0-9!#$%&'*+\/=?^_`{|}~-]+)*|\"(?:[\x01-\x08\x0b\x0c\x0e-\x1f\x21\x23-\x5b\x5d-\x7f]|\\[\x01-\x09\x0b\x0c\x0e-\x7f])*\")@(?:(?:[a-z0-9](?:[a-z0-9-]*[a-z0-9])?\.)+[a-z0-9](?:[a-z0-9-]*[a-z0-9])?|\[(?:(?:25[0-5]|2[0-4][0-9]|[01]?[0-9][0-9]?)\.){3}(?:25[0-5]|2[0-4][0-9]|[01]?[0-9][0-9]?|[a-z0-9-]*[a-z0-9]:(?:[\x01-\x08\x0b\x0c\x0e-\x1f\x21-\x5a\x53-\x7f]|\\[\x01-\x09\x0b\x0c\x0e-\x7f])+)\])"
```

```python
import re
def find_emails_in_file(file_name):
    with open(file_name, encoding = 'utf-8') as my_file:
        
        f = my_file.read()
        found = re.findall(regex, f )
    return found
```

```python
print(
    'Funkcja find_emails_in_file znalazła',
    len(find_emails_in_file('data.txt')),
    'adresów email (oczekiwano 40)'
)

if find_emails_in_file('data.txt')[0] == 'kszpyra@gmail.com':
    print('Pierwszy znaleziony adres to kszpyra@gmail.com - zgodnie z oczekiwaniami')
else:
    print(
        f'Pierwszy adres zwrócony przez funkcję to '
        f'{find_emails_in_file("data.txt")[0]!r} - oczekiwano kszpyra@gmail.com'
    )

if find_emails_in_file('data.txt')[-1] == 'n.strzelczak@wp.pl':
    print('Ostatni znaleziony adres to n.strzelczak@wp.pl - zgodnie z oczekiwaniami')
else:
    print(
        f'Ostatni adres zwrócony przez funkcję to '
        f'{find_emails_in_file("data.txt")[-1]!r} - oczekiwano n.strzelczak@wp.pl'
    )
```

Funkcja `strip_non_letters(text)`, która przyjmuje tekst, oraz zwraca go pozostawiając tylko małe i duże litery, bez akcentów i znaków diakrytycznych.

```python
import re

def strip_non_letters(text):
    return re.sub(r'[^a-zA-Z ]', '', text)
```

```python
if strip_non_letters('Zażółć gęślą jaźń') == 'Za gl ja':
    print('Funkcja poprawnie pozbywa się polskich znaków')
else:
    print('Dla frazy "Zażółć gęślą jaźń" oczekiwano wyniku "Za gl ja",')
    print('ale funkcja zwróciła', repr(strip_non_letters('Zażółć gęślą jaźń')))
```
