```python
def multiply(a, b):
    return a * b
```

```python
print(multiply(3, 8))
```

```python
def power(base, exponent):
    return base ** exponent
```

```python
power(3, 3)
```

```python
def root (base, degree):
    return base ** (1 / degree)
```

```python
root(16, 4)
```

```python
def usd_to_pln(usd):
    return usd * 3.95
```

```python
usd_to_pln(10)
```

```python
def is_even(number):
    return not number % 2
    
```

```python
is_even(78)
```

```python
def is_odd(number):
    return not(is_even(number))
```

```python
is_odd(98)
```
