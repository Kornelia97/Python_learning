
Loads file

```python
import pandas as pd

df = pd.read_csv(
    r'../Data/ceny_produktow_renamed.csv',
    sep=';',
    decimal='.'
)
df.head()
```

In column **date** there are data from 1888 - '1888-0', change value to 1999-1

```python
df.loc[df['date'] == '1888-0', 'date'] = '1999-1'
```

```python
df.loc[df['date'] == '1888-0'].head()
```

```python
df.loc[df['date'] == '1999-1'].head()
```

In column **date** there are data from 2099 - '2099-13', change value to 2019-1

```python
df.loc[df['date'] == '2099-13', 'date'] = '2019-1'
```

```python
df.loc[df['date'] == '2099-13'].head()
```

 In column **product_types** appeared a spelling error.

```python
df['product_types'].unique()
```

```python
df.loc[df['product_types'] == 'jaja kóże śfierze - za 666szt.', 'product_types'] = 'jaja kurze świeże - za 10szt.'
```

Checks whether the data has been changed correctly.

   Converts values in `EUR` to `zł` using an exchange rate of 4.15.

```python
df.loc[df['currency'] == 'EUR', 'value'] *= 4.15
df.loc[df['currency'] == 'EUR', 'currency'] = 'zł'
```

```python
df.loc[df['currency'] == 'EUR'].head()
```

```python
df.loc[df['currency'] == 'zł'].head()
```

Removes from the dataset the rows where the product price is 3000.

```python
df = df.loc[df['value'] != 3000]
```
