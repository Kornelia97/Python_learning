# Adding a column

Create a new column `product` using `product_types` and `product_line`:

- Check whether the `product_types` and `product_line` columns complement each other (i.e., a null value in one column implies a non-null value in the other).

```python
# Returns True if the columns perfectly complement each other in each row
do_they_complement = (df['product_types'].notna() ^ df['product_line'].notna()).all()
print(do_they_complement)
```

- Creates a new column named 'product' using the values ​​from the 'product_types' column.
- Identifies the non-null values ​​in the 'product_line' column and assign them to the 'product' column.
- Verifies, that all values ​​in the 'product' column are non-null.

```python
# 1. Create a new column 'product' based on 'product_types'
df['product'] = df['product_types']

# 2. Fill 'product' with non-null values ​​from 'product_line'
mask = df['product_line'].notna()
df.loc[mask, 'product'] = df.loc[mask, 'product_line']

# 3. Check if all values ​​in the 'product' column are non-null (returns True or False)
all_not_null = df['product'].notna().all()
print(all_not_null)
```

- Removes duplicate rows from the table.

```python
df= df.drop_duplicates()
```
