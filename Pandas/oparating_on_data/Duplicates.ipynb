# Duplicates

Uploads **ceny_produktow_renamed.csv**

```python
import pandas as pd

df = pd.read_csv(
    r'../Data/ceny_produktow_renamed.csv',
    sep=';',
    decimal='.'
)
df.head()
```

Checks how many rows are duplicated.

```python
df.duplicated().sum()
```

Finds which rows has been duplicated.

```python
duplicates = df.duplicated()
df.loc[duplicates]
```

Using drop_duplicates, remove all duplicates from the dataset(we assume that if a row has a duplicate, it is incorrect and should be removed) – seves the result to a new frame.

```python
df_without_duplicates = df.drop_duplicates(keep=False)
```
