# Highest and lowest price

Using `sort_values` and the data from the file _ceny_produktow_renamed.csv_, answer the questions below — use the data aggregated for all of Poland (`province=='POLSKA'`).:

1. Which product reached the lowest price in PLN throughout its entire history? (remember to filter out values ​​equal to 0),
2. Which product reached the highest price in PLN throughout its entire history? (remember to exclude values ​​equal to 3000),
3. Sort the data according to the following criteria:
- by column - **product_types** - descending
- by column - **value** - ascending

```python
import pandas as pd

df_raw = pd.read_csv(
    r'../Data/ceny_produktow_renamed.csv',
    sep=';',
    decimal='.'
)
```

```python
df = df_raw.loc[(df_raw['value'] != 0) & (df_raw['value'] != 3000) & (df_raw['province'] == 'POLSKA')]
```

1. Lowest value

```python
df_sort_asc = df.sort_values(by = 'value', ascending = True) 
df_sort_asc['value'].head(1)
```

2. Highest value

```python
df_sort_desc = df.sort_values(by ='value', ascending = False) 
df_sort_desc['product_line'].head(1)
```

3. Sorted

```python
df_sorted = df.sort_values(by=['product_types', 'value'], ascending = [False, True])
df_sorted.head()
```
