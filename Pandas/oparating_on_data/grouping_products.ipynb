# Grouping products

Using data from the file **ceny_produktow_cleaned.csv** for the provinces.

```python
import pandas as pd

df = pd.read_csv(
    r'../Data/ceny_produktow_cleaned.csv',
    sep=';',
    decimal='.'
)
```

- Prepares a data frame using provinces.

```python
df_province = df.loc[df['province'] != 'POLSKA']
```

- Average monthly price of individual products.

```python
df_by_product = df_province.groupby(by = ['product', 'date'])
```

```python
df_by_product['value'].mean()
```

- Which product exhibited the greatest price volatility over the years.

```python
df_province.groupby('product')['value'].std().max()
```
