```python
import pandas as pd

df = pd.read_csv(
    r'../Data/ceny_produktow_cleaned.csv',
    sep = ';',
    decimal='.'
)
```

- Aggregates the data for each product on a monthly basis, then calculate the statistics `min, max, median, mean, std` for the prices(excludes nationwide data from the analysis)

```python
df_province = df.loc[df['province'] != 'POLSKA']
```

```python
df_province.groupby(['product', 'date'])['value'].agg(['min', 'max', 'median', 'mean', 'std'])
```

- A loop that will calculate the above values ​​for the individual voivodeships.

```python
df_by_province = df.groupby('province')
```

```python
for(key, df_province) in df_by_province:
    print(key)
    print(df_province['value'].agg(['min', 'max', 'median', 'mean', 'std']))
```
