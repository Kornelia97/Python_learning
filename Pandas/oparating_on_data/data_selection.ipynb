```python
import pandas as pd

df = pd.read_csv(
  r'..\Data\ceny_produktow_renamed.csv',
  sep=';',
  decimal='.'
)
```

Average price of a carrot and pea mix per 1 kg in Poland in 2015

```python
df_carrot = df.query("date >= '2015-1' and date <= '2015-12' and product_types == 'mieszanka mrożona marchew i groszek - za 1kg' and province == 'POLSKA'").head()
df_carrot['value'].mean()
```

Average price of apple juice in the years 2016–2018 in `MAZOWIECKIE` Voivodeship

```python
df_apple_juice = df.query("date >= '2016-1' and date <= '2018-12' and product_types == 'sok jabłkowy, w opakowaniu kartonowym - za 1l' and province == 'MAZOWIECKIE'")
df_apple_juice['value'].mean()
```

Average price of tomato concentrate in the years 2003-2015

```python
df_concentrate = df.query("date >= '2003-1' and date <= '2015-12' and product_types == 'koncentrat pomidorowy 30% - za 1kg' and province == 'DOLNOŚLĄSKIE'")
df_concentrate['value'].mean()
```

Average looks suspicious so I need to try other statistics

```python
df_concentrate['value'].describe()
```

It looks like there is a lot of `0` values, max value `3000` looks suspicious as well, so I think we should cut them.

```python
df_conc = df_concentrate.loc[~(df_concentrate['value'].isin([0,3000]))]
df_conc['value'].describe()
```
