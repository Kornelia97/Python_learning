

**1.** Kod, który wyświetli treść pliku `todo.json`za pomocą funkcji `print`, w formie:
```
<IMIĘ PODANE PRZEZ UŻYTKOWNIKA>, twoje bojowe zadanie na dziś to <ZADANIE #1>
<IMIĘ PODANE PRZEZ UŻYTKOWNIKA>, twoje bojowe zadanie na dziś to <ZADANIE #2>
...

````python
import json

with open('todo.json', encoding = 'utf-8') as file:
    data = json.loads(file.read())

    for task in data['tasks']:
        print(f"{data['owner']}, twoje bojowe zadanie na dziś to: {task} ")

````


**2.** Kod, który zapyta użytkownika o imię, a następnie utworzy plik `todo.json` o treści:
```json
{
  "owner": "<IMIĘ PODANE PRZEZ UŻYTKOWNIKA>",
  "tasks": []
}
```
Jeśli plik już istnieje, powinien zostać nadpisany.

````python
import json

your_name = input("Podaj swoje imię: ")

with open('todo.json', 'w', encoding = 'utf-8') as file:
    
    data ={
  "owner": your_name,
  "tasks": []
}

    file.write(json.dumps(data, indent = 4))



````

**3.** Kod, który zapyta o treść nowego zadania, wczyta obecną zawartość pliku, a następnie dopisze je jako string do listy `"tasks"` w pliku. Uruchamiając tę komórkę Jupytera kilka razy zapiszesz kilka zadań na liście:
```json
{
  "owner": "Jan",
  "tasks": ["kupić chleb", "kupić masło", "kupić mleko", "dolać oleju do Hondy"]
}
```

Jeśli plik już istnieje, powinien zostać nadpisany.

````python
import json

new_task = input("Podaj nowe zadanie: ")

with open('todo.json', 'r', encoding = "utf-8") as file:

    data = json.loads(file.read())

    data['tasks'].append(new_task)

    print(data)

with open('todo.json', 'w', encoding = "utf-8") as file:
    file.write(json.dumps(data, indent = 4, ensure_ascii = False))



````
