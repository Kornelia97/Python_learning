Open JSON file and deserialization.

```python
import json

with open('clients.json') as file:
    data = json.loads(file.read())
    clients = (data['clients'])
```

```python
print("Lista imion i nazwisk kobiet:")

women_lst = []
    
for client in clients:
    if client['gender'] == 'female':
        women_lst.append(client['name'])
print(women_lst)

```

```python
print("Lista imion i nazwisk mężczyzn:")

men_lst = []

for client in clients:
    if client['gender'] == 'male':
        men_lst.append(client['name'])
print(men_lst)
```

```python
print("Lista aktywnych klientów:")

active_clients = []
for client in clients:
    if client['isActive']:
        active_clients.append(client['name'])
print(active_clients)
```

```python
print("Lista klientów z tagiem `adipisicing`:")

adipisicing_clients = []

for client in clients:
    if 'adipisicing' in client['tags']:
        adipisicing_clients.append(client['name'])
print(adipisicing_clients)
```

```python
print("Lista klientów o zielonych oczach:")

green_eyed_clients = []

for client in clients:
    if client['eyeColor'] == 'green':
        green_eyed_clients.append(client['name'])

print(green_eyed_clients)
```

```python
import re

print("Lista klientów z polskim numerem telefonu:")

regex = r"\+48[0-9 ]{9,12}"

polish_phone_clients = []

for client in clients:
    if re.match(regex, client['phone']):
        polish_phone_clients.append(client['name'])
print(polish_phone_clients)
```
